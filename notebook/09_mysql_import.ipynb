```python
import pandas as pd
import mysql.connector
from mysql.connector import Error
from sqlalchemy import create_engine
```

```python
HOST = "localhost"
USER = "root"
PASSWORD = "suthar111"
DATABASE = "movieverse_db"
```

```python
engine = create_engine(
    f"mysql+mysqlconnector://{USER}:{PASSWORD}@{HOST}/{DATABASE}"
)

print("Connected Successfully")
```

```python
movies = pd.read_csv(
    "../dataset/processed/movieverse_master_dataset.csv"
)

movies.head()
```

```python
movies = movies[
    [
        "movie_id",
        "title",
        "release_year",
        "tagline",
        "overview",
        "runtime_minutes",
        "avg_rating",
        "poster_url"
    ]
]
```

```python
movies.to_sql(
    "movies",
    engine,
    if_exists="replace",
    index=False,
    chunksize=5000,
    method="multi"
)

print("Movies Imported")
```

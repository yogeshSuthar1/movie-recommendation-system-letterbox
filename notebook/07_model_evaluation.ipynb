```python
import pandas as pd
import joblib
import time
```

```python
movies = pd.read_pickle("../models/movies.pkl")

knn = joblib.load("../models/recommendation_model.pkl")

movie_index = joblib.load(
    "../models/movie_index.pkl"
)
tfidf_matrix = joblib.load("../models/tfidf_matrix.pkl")
```

```python
def recommend(

    movie_name,

    start=0,

    limit=20

):

    movie_name = movie_name.lower()

    if not movie_name:
        return "Movie not found."

    if movie_name not in movie_index:
        return "Movie not found."

    idx = movie_index[movie_name]

    distances, indices = knn.kneighbors(

        tfidf_matrix[idx],

        n_neighbors=start+limit+1

    )

    indices = indices[0][1:]

    distances = distances[0][1:]

    recommendations = movies.iloc[indices].copy()
    recommendations["similarity"] = 1 - distances

    return recommendations.iloc[start:start+limit][

        [

            "title",

            "release_year",

            "avg_rating",

            "genres",

            "similarity",

            "poster_url"

        ]

    ]
```

```python
test_movies = [
    "Interstellar",
    "Inception",
    "The Dark Knight",
    "Parasite",
    "Barbie",
    "Avengers: Endgame",
    "Dangal",
    "3 Idiots",
    "The Godfather",
    "Titanic"
]
```

```python
for movie in test_movies:

    print("="*80)

    print(f"Movie : {movie}\n")

    print(recommend(movie, limit=10))
```

```python
times = []

for movie in test_movies:

    start = time.time()

    recommend(movie)

    end = time.time()

    times.append(end-start)
```

```python
print(f"Average Time : {sum(times)/len(times):.4f} sec")
print(f"Fastest : {min(times):.4f} sec")
print(f"Slowest : {max(times):.4f} sec")
```

```python
test_movies = [
    "Interstellar",
    "Inception",
    "The Dark Knight",
    "Parasite",
    "Barbie",
    "Avengers: Endgame",
    "Dangal",
    "3 Idiots",
    "The Godfather",
    "Titanic",
    "abcdef"
]

success = 0
failed = 0

for movie in test_movies:

    result = recommend(movie)

    if isinstance(result, str):
        failed += 1
    else:
        success += 1

print(f"Successful : {success}")
print(f"Failed : {failed}")
print(f"Success Rate : {(success/(success+failed))*100:.2f}%")
```

```python
import pandas as pd
import numpy as np
```

### missing values

```python
missing = (
    movies.isnull()
    .sum()
    .sort_values(ascending=False)
)

missing
```

```python
missing_percentage = (
    movies.isnull().mean() * 100
).round(2)

missing_percentage.sort_values(ascending=False)
```

### title clean


```python
movies["title"] = (
    movies["title"]
    .astype(str)
    .str.strip()
)
```

### overview

```python
movies["overview"] = movies["overview"].fillna(
    "Overview not available."
)
```

### tagline

```python
movies["tagline"] = movies["tagline"].fillna("")
```

### cast

```python
movies["main_cast"] = movies["main_cast"].fillna(
    "Cast information unavailable"
)
```

### countries

```python
movies["countries"] = movies["countries"].fillna(
    "Unknown"
)
```

### producer

```python
movies["producer"] = movies["producer"].fillna(
    "Unknown"
)
```

### studio

```python
movies["studios"] = movies["studios"].fillna(
    "Unknown"
)
```

### language

```python
movies["languages"] = movies["languages"].fillna(
    "Unknown"
)
```

### datatype of data and time 

```python
movies["release_year"] = movies["release_year"].astype("Int64")

movies["runtime_minutes"] = movies["runtime_minutes"].astype("Int64")
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

### check empty string 

```python
for col in movies.select_dtypes(include="object").columns:
    empty = (movies[col].str.strip() == "").sum()
    print(f"{col}: {empty}")
```

### release year validation 

```python
print("Minimum Year:", movies["release_year"].min())
print("Maximum Year:", movies["release_year"].max())
```

### save clean dataset 

```python
movies.to_csv(
    "../dataset/processed/movieverse_master_dataset_clean.csv",
    index=False
)

print("Clean Dataset Saved Successfully!")
```

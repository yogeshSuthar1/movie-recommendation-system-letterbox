## Genres Table

```python
genres_table = (
    movies["genres"]
    .dropna()
    .str.split(" | ", regex=False)
    .explode()
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
    .to_frame(name="genre_name")
)

genres_table.insert(0, "genre_id", range(1, len(genres_table) + 1))

genres_table.head()
```

```python
genres_table.shape
```

##  Save Genres Table

```python
genres_table.to_csv(
    "../dataset/database/genres.csv",
    index=False
)

print("Genres Table Saved")
```

## Genre Mapping Dictionary

```python
genre_map = dict(
    zip(
        genres_table["genre_name"],
        genres_table["genre_id"]
    )
)

list(genre_map.items())[:5]
```

## Create Movie Genres Table

```python
movie_genres = (
    movies[
        ["movie_id", "genres"]
    ]
    .dropna()
)

movie_genres["genres"] = movie_genres["genres"].str.split(" | ", regex=False)

movie_genres = movie_genres.explode("genres")

movie_genres["genre_id"] = movie_genres["genres"].map(genre_map)

movie_genres = movie_genres[
    ["movie_id", "genre_id"]
]
```

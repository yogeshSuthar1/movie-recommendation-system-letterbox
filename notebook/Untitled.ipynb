```python
import pandas as pd
```

```python
movies = pd.read_csv("../dataset/processed/movieverse_master_dataset.csv")

print(movies["runtime_minutes"].max())
```

```python
movies.loc[movies["runtime_minutes"] > 300,
           ["movie_id", "title", "runtime_minutes"]]
```

```python
movies = movies[
    (movies["runtime_minutes"].isna()) |
    ((movies["runtime_minutes"] >= 60) &
     (movies["runtime_minutes"] <= 300))
]
```

```python
movies.to_csv(
    "../dataset/processed/movieverse_master_dataset.csv",
    index=False
)
```

```python
import pandas as pd

actors = pd.read_csv("../dataset/processed/actors.csv")

print(actors.shape)
print(actors["actor_id"].max())
```

```python
import pandas as pd

movie_actors = pd.read_csv("../dataset/processed/movie_actors.csv")

print("Max Actor ID:", actors["actor_id"].max())
print("Max Movie Actor ID:", movie_actors["actor_id"].max())
```

```python
missing = set(movie_actors["actor_id"]) - set(actors["actor_id"])

print(len(missing))
print(list(missing)[:20])
```

```python
duplicates = actors[actors.duplicated(subset=["actor_name"], keep=False)]

print("Duplicate actor names:", len(duplicates))
print(duplicates.head(20))
```

```python
actors = actors.drop_duplicates(subset=["actor_name"]).reset_index(drop=True)

actors["actor_id"] = range(1, len(actors) + 1)

actors.to_csv(
    "../dataset/processed/actors.csv",
    index=False
)

print(actors.shape)
```

```python
dup = actors[actors.duplicated("actor_name", keep=False)]

print(dup.shape)
print(dup[dup["actor_name"].str.contains("Weird Al", case=False, na=False)])
```

```python
actors_work = actors[["id", "name", "role"]].copy()

actors_work["name"] = actors_work["name"].fillna("")
actors_work["role"] = actors_work["role"].fillna("")

# Har movie ke liye sirf pehle 7 actors lo
actors_work["rank"] = actors_work.groupby("id").cumcount()

actors_top7 = actors_work[actors_work["rank"] < 7].copy()

actors_top7["cast_item"] = (
    actors_top7["name"].str.strip() + " as " + actors_top7["role"].str.strip()
)

actors_clean = (
    actors_top7.groupby("id", sort=False)["cast_item"]
    .agg(" | ".join)
    .reset_index(name="main_cast")
)
```

```python
genres_clean = (
    genres.groupby("id")["genre"]
    .apply(lambda x: " | ".join(sorted(set(x))))
    .reset_index(name="genres")
)
```

```python
movies_clean = movies.copy()

# Sirf runtime 60 min se kam wali movies remove karo
movies_clean = movies_clean[
    (movies_clean["minute"].isna()) |
    (movies_clean["minute"] >= 60)
].copy()

movies_clean = movies_clean.rename(columns={
    "id": "movie_id",
    "name": "title",
    "date": "release_year",
    "minute": "runtime_minutes",
    "rating": "avg_rating",
    "description": "overview"
})

movies_clean["release_year"] = movies_clean["release_year"].astype("Int64")
movies_clean["runtime_minutes"] = movies_clean["runtime_minutes"].astype("Int64")
movies_clean["avg_rating"] = movies_clean["avg_rating"].astype("float")

print(movies_clean.shape)
movies_clean.head()
```

```python
final_movies = movies_clean.merge(
    genres_clean,
    left_on="movie_id",
    right_on="id",
    how="left"
)

final_movies = final_movies.merge(
    actors_clean,
    left_on="movie_id",
    right_on="id",
    how="left",
    suffixes=("", "_actors")
)

final_movies = final_movies.merge(
    posters_clean,
    left_on="movie_id",
    right_on="id",
    how="left",
    suffixes=("", "_posters")
)

final_movies = final_movies.merge(
    director_clean,
    left_on="movie_id",
    right_on="id",
    how="left",
    suffixes=("", "_director")
)

final_movies = final_movies.merge(
    producer_clean,
    left_on="movie_id",
    right_on="id",
    how="left",
    suffixes=("", "_producer")
)

# Extra id columns remove
final_movies = final_movies.loc[:, ~final_movies.columns.str.startswith("id")]

print(final_movies.shape)
```
